```python
import numpy as np

# 1. Remove duplicate rows if any
df = df.drop_duplicates()

# 2. Handle missing values
# For columns with mostly numeric data stored as strings, we will convert them to lists of floats and handle missing values accordingly


# Helper function to convert comma-separated strings to lists of floats
def str_to_float_list(s):
    if pd.isnull(s):
        return np.nan
    try:
        return [float(x) for x in str(s).split(',') if x.strip() != '']
    except Exception:
        return np.nan

# Convert relevant columns to lists of floats
for col in ['A Current', 'A Voltage', 'B Current', 'B Voltage']:
    df[col] = df[col].apply(str_to_float_list)

# Optionally, drop rows where any of these columns are NaN (if you want only complete data)
df = df.dropna(subset=['A Current', 'A Voltage', 'B Current', 'B Voltage'])

# 3. Convert 'Time' column to datetime
df['Time'] = pd.to_datetime(df['Time'], errors='coerce')

# Drop rows with invalid or missing datetime
df = df.dropna(subset=['Time'])

# 4. Reset index after cleaning
df = df.reset_index(drop=True)

# 5. (Optional) Standardize string columns (strip whitespace, upper/lowercase if needed)
for col in ['Site Name', 'Point Machine Name', 'Direction', 'Type of A', 'Type of B']:
    df[col] = df[col].astype(str).str.strip()

# 6. Ensure 'Polling of A' and 'Polling of B' are integers
df['Polling of A'] = pd.to_numeric(df['Polling of A'], errors='coerce').astype('Int64')
df['Polling of B'] = pd.to_numeric(df['Polling of B'], errors='coerce').astype('Int64')

# 7. Drop any remaining rows with missing critical values
df = df.dropna(subset=['Site Name', 'Point Machine Name', 'Direction', 'Type of A', 'Type of B', 'Polling of A', 'Polling of B'])

# Now df is cleaned and ready for analysis and algorithms
df.head()
```

```python
# Find rows where 'A Current' or 'B Current' columns contain lists with only 0.0 values

def is_all_zeros(lst):
    return isinstance(lst, list) and all(x == 0.0 for x in lst) and len(lst) > 0

a_current_zeros = df['A Current'].apply(is_all_zeros)
b_current_zeros = df['B Current'].apply(is_all_zeros)

# Rows where either 'A Current' or 'B Current' is all zeros
rows_with_all_zeros = df[a_current_zeros | b_current_zeros]

# Display the relevant rows
rows_with_all_zeros
```

```python
# Analyze rows with all zeros in 'A Current' or 'B Current' (rows_with_all_zeros)

# 1. Distribution by Site
site_counts = rows_with_all_zeros['Site Name'].value_counts()
print("Distribution by Site Name:\n", site_counts)

# 2. Distribution by Direction
direction_counts = rows_with_all_zeros['Direction'].value_counts()
print("\nDistribution by Direction:\n", direction_counts)

# 3. Time analysis: Are these events clustered in time?
import matplotlib.pyplot as plt

# Plot histogram of occurrence times
plt.figure(figsize=(10, 4))
rows_with_all_zeros['Time'].hist(bins=30)
plt.title("Distribution of 'All Zeros' Events Over Time")
plt.xlabel("Time")
plt.ylabel("Count")
plt.tight_layout()
plt.show()

# 4. Check if these events are consecutive or have any time pattern
rows_with_all_zeros_sorted = rows_with_all_zeros.sort_values('Time')
time_diffs = rows_with_all_zeros_sorted['Time'].diff().dropna()
print("\nTime differences between consecutive 'all zeros' events (in minutes):")
print(time_diffs.dt.total_seconds() / 60)

# 5. Optional: Check if both 'A Current' and 'B Current' are zero in the same row
def both_all_zeros(row):
    return isinstance(row['A Current'], list) and all(x == 0.0 for x in row['A Current']) and \
           isinstance(row['B Current'], list) and all(x == 0.0 for x in row['B Current'])

both_zero_rows = rows_with_all_zeros[rows_with_all_zeros.apply(both_all_zeros, axis=1)]
print(f"\nRows where BOTH 'A Current' and 'B Current' are all zeros: {len(both_zero_rows)}")
```
