```python
import pandas as pd
import numpy as np

# --- Step 1: Load Data ---
df = pd.read_csv("/Users/anamikasaroha/Energy7/week_10/filtered_NFR.csv")
```

```python
import matplotlib.pyplot as plt

# Filter for Polling of A = 20 and Polling of A = 100
df_poll_20 = df[df["Polling of A"] == 20]
df_poll_100 = df[df["Polling of A"] == 100]

# Sample 20 rows from each
sample_20 = df_poll_20.sample(n=20, random_state=42)
sample_100 = df_poll_100.sample(n=20, random_state=42)

# Plot for Polling of A = 20
plt.figure(figsize=(12, 5))
for idx, row in sample_20.iterrows():
    signal = [float(x.strip()) for x in row["A Current"].split(",")]
    plt.plot(signal, alpha=0.7)
plt.title("A Current Signals (20 samples, Polling of A = 20)")
plt.xlabel("Sample Index")
plt.ylabel("Current")
plt.show()

# Plot for Polling of A = 100
plt.figure(figsize=(12, 5))
for idx, row in sample_100.iterrows():
    signal = [float(x.strip()) for x in row["A Current"].split(",")]
    plt.plot(signal, alpha=0.7)
plt.title("A Current Signals (20 samples, Polling of A = 100)")
plt.xlabel("Sample Index")
plt.ylabel("Current")
plt.show()
```

```python
# Plot for Polling of B = 20
plt.figure(figsize=(12, 5))
for idx, row in sample_20.iterrows():
    signal = [float(x.strip()) for x in row["B Current"].split(",")]
    plt.plot(signal, alpha=0.7)
plt.title("B Current Signals (20 samples, Polling of B = 20)")
plt.xlabel("Sample Index")
plt.ylabel("Current")
plt.show()

# Plot for Polling of B = 100
plt.figure(figsize=(12, 5))
for idx, row in sample_100.iterrows():
    signal = [float(x.strip()) for x in row["B Current"].split(",")]
    plt.plot(signal, alpha=0.7)
plt.title("B Current Signals (20 samples, Polling of B = 100)")
plt.xlabel("Sample Index")
plt.ylabel("Current")
plt.show()
```

```python
# Create dataset with Polling of A and B both equal to 100
df_poll_100 = df[(df["Polling of A"] == 100) & (df["Polling of B"] == 100)]

# Create dataset with Polling of A and B both equal to 20
df_poll_20 = df[(df["Polling of A"] == 20) & (df["Polling of B"] == 20)]
```

```python
df_poll_100.to_csv("NFR_poll_100.csv", index=False)
df_poll_20.to_csv("NFR_poll_20.csv", index=False)
```
